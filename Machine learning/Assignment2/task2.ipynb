## Sub-task 2: Dimensionality Reduction

This notebook presents the application of Sparse PCA to a dataset containing price and length-based attributes. A model with two components is fitted on both raw and scaled data. The resulting components are printed and analyzed.

To evaluate the results, the impact of data scaling on Sparse PCA is discussed, and the findings are compared against a potential latent variable model, highlighting the importance of scaling in uncovering meaningful structures.

```python
#import libraries
import pandas as pd
from sklearn.decomposition import SparsePCA
from sklearn.preprocessing import StandardScaler
```

```python
# Load the dataset
df = pd.read_csv('cars_reduced.csv', header=None, names=['retail_price', 'dealer_price', 'wheel_base', 'length'])
print(len(df))
```

```python
df.head(10)
```

```python
# Apply Sparse PCA with alpha = 5
sparse_pca_raw = SparsePCA(n_components=2, alpha=5, random_state=42)
sparse_pca_raw.fit(df)

# Print components
print("Sparse PCA components on raw data:")
print(pd.DataFrame(sparse_pca_raw.components_, columns=df.columns))
```

```python
# Standardize the data
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)
```

```python
# Apply Sparse PCA again on scaled data
sparse_pca_scaled = SparsePCA(n_components=2, alpha=5, random_state=42)
sparse_pca_scaled.fit(df_scaled)

# Print components
print("Sparse PCA components on scaled data:")
print(pd.DataFrame(sparse_pca_scaled.components_, columns=df.columns))
```

### Insight gained:

When Sparse PCA is applied to the raw data, the components are dominated by the price variables (retail price and dealer price), while the length variables (wheel base and length) have almost no contribution. This happened because PCA (and Sparse PCA) are sensitive to the scale of the input variables and in the raw data, price variables likely had much larger numerical ranges compared to length-based variables.

After scaling the data, the Sparse PCA components became much cleaner and interpretable:

The first component captures retail price and dealer price equally.

The second component captures wheel base and length equally.

This scaled data result better reflects the relationships between the variables and successfully separates the price aspects and length aspects into different components, supporting the latent variable model.
Thus, the result after scaling is more meaningful and most likely to capture the true structure of the dataset.
